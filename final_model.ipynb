```python
def ask_symptoms(symptoms):
    answers = {}

    for symptom in symptoms:
        answer = input(f"Do you have {symptom}? (yes/no): ")

        if answer.lower() == "yes":
            answers[symptom] = 1
        else:
            answers[symptom] = 0

    return answers


print("Select your problem:")
print("1. Skin Problem")
print("2. Respiratory Problem")
print("3. Fever / Infection")
print("4. Neurological Problem")
print("5. Muscle / Joint Problem")
print("6. Digestive Problem")
print("7. Urinary Problem")
print("8. Heart / Circulation Problem")
print("9. Eye Problem")
print("10. Liver / Jaundice Problem")
print("11. Mental / Emotional Problem")
print("12. Weight / Metabolism Problem")
print("13. Anal / Bowel Problem")
print("14. Thyroid / Neck Problem")
print("15. Mouth / Lip Problem")
print("16. Swelling / Fluid Problem")

user = int(input("Enter your problem number: "))


if user == 1:
    category_answers = ask_symptoms(skin_symptoms)

elif user == 2:
    category_answers = ask_symptoms(respiratory_symptoms)

elif user == 3:
    category_answers = ask_symptoms(fever_infection_symptoms)

elif user == 4:
    category_answers = ask_symptoms(neurological_symptoms)

elif user == 5:
    category_answers = ask_symptoms(muscle_joint_symptoms)

elif user == 6:
    category_answers = ask_symptoms(digestive_symptoms)

elif user == 7:
    category_answers = ask_symptoms(urinary_symptoms)

elif user == 8:
    category_answers = ask_symptoms(heart_circulation_symptoms)

elif user == 9:
    category_answers = ask_symptoms(eye_symptoms)

elif user == 10:
    category_answers = ask_symptoms(liver_jaundice_symptoms)

elif user == 11:
    category_answers = ask_symptoms(mental_emotional_symptoms)

elif user == 12:
    category_answers = ask_symptoms(metabolic_symptoms)

elif user == 13:
    category_answers = ask_symptoms(anal_bowel_symptoms)

elif user == 14:
    category_answers = ask_symptoms(thyroid_neck_symptoms)

elif user == 15:
    category_answers = ask_symptoms(mouth_lip_symptoms)

elif user == 16:
    category_answers = ask_symptoms(swelling_fluid_symptoms)

else:
    print("Invalid choice")


common_answers = ask_symptoms(common_symptoms)

all_answers = {**category_answers, **common_answers}


index = [0] * 132

for symptom, value in all_answers.items():
    i = x.columns.get_loc(symptom)
    index[i] = value
```
